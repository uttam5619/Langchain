```python
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
api = os.getenv('OPENAI_KEY')
```

```python

llm = ChatOpenAI(
    api_key = api,
    model = 'gpt-4.1-nano',
    temperature =0
)
```

```python
@tool
def multiply(a:int, b:int) -> int:
    """ 
        multiply tool multiplies two nuumbers and return the result
        args: a:int , b:int
    """
    return a*b
```

```python
result = multiply.invoke({'a':5,'b':6})
print(result)
```

```python
@tool
def get_word_len(word:str)->str:
    """get_word_len takes a sting as input and returns its length"""
    return len(word)
```

```python
get_word_len.args
```

```python
get_word_len.invoke({'word':'hello'})
```

```python
llm_with_tools = llm.bind_tools([multiply,get_word_len])
```

```python
response = llm_with_tools.invoke('What is 3 multiplesd 4')
print(response)
print(response.tool_calls[0])
```
